# **Competition Code**

## 1. Preprocessing Steps

```python
### Add the code related to the preprocessing steps in cells inside this section
import pandas as pd
import numpy as np
import re
from pathlib import Path
from sklearn.model_selection import train_test_split

kaggle_path = Path("/kaggle/input/competitions/2026-dm-lab-2-private-competition")
data_path = kaggle_path if kaggle_path.exists() else Path(".")

posts = pd.read_json(data_path / "final_posts.json")
data_info = pd.read_csv(data_path / "data_identification.csv")
emotion = pd.read_csv(data_path / "emotion.csv")
sample = pd.read_csv(data_path / "sampleSubmission.csv")

df = posts.merge(data_info, on="post_id").merge(emotion, on="post_id", how="left")
train_df = df[df["set"] == "train"].copy()
test_df = df[df["set"] == "test"].copy()

print("Train:", train_df.shape)
print("Test:", test_df.shape)
print("Emotion labels:")
print(train_df["label"].value_counts())
print("Missing training texts:", train_df["text"].isna().sum())
print("Missing training labels:", train_df["label"].isna().sum())
print("Missing test texts:", test_df["text"].isna().sum())

def clean_text(text):
    text = str(text).lower()
    text = re.sub(r"http\S+|www\S+", " ", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()

train_df["clean_text"] = train_df["text"].apply(clean_text)
test_df["clean_text"] = test_df["text"].apply(clean_text)

X_train, X_val, y_train, y_val = train_test_split(
    train_df["clean_text"], train_df["label"],
    test_size=0.2, random_state=42, stratify=train_df["label"]
)
print("Training split:", len(X_train))
print("Validation split:", len(X_val))
```

## 2. Feature Engineering Steps

```python
### Add the code related to the feature engineering steps in cells inside this section
from sklearn.feature_extraction.text import TfidfVectorizer
from scipy.sparse import hstack, csr_matrix
from sklearn.preprocessing import normalize
word_vectorizer = TfidfVectorizer(ngram_range=(1, 2), min_df=2, max_features=20000)
char_vectorizer = TfidfVectorizer(analyzer="char", ngram_range=(2, 4), min_df=2, max_features=30000)
X_train_word = word_vectorizer.fit_transform(X_train)
X_val_word = word_vectorizer.transform(X_val)
X_train_char = char_vectorizer.fit_transform(X_train)
X_val_char = char_vectorizer.transform(X_val)
X_train_tfidf = hstack([X_train_word, X_train_char])
X_val_tfidf = hstack([X_val_word, X_val_char])
print("Word TF-IDF:", X_train_word.shape)
print("Character TF-IDF:", X_train_char.shape)
print("Combined TF-IDF:", X_train_tfidf.shape)

# MiniLM Sentence Embeddings
from sentence_transformers import SentenceTransformer
minilm = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")
X_train_mini = normalize(minilm.encode(X_train.tolist(), batch_size=64, show_progress_bar=True))
X_val_mini = normalize(minilm.encode(X_val.tolist(), batch_size=64, show_progress_bar=True))
X_train_v5 = hstack([X_train_word, X_train_char, csr_matrix(X_train_mini * 0.5)])
X_val_v5 = hstack([X_val_word, X_val_char, csr_matrix(X_val_mini * 0.5)])
print("MiniLM:", X_train_mini.shape)
print("Fusion:", X_train_v5.shape)
```

## 3. Model Implementation Steps

```python
### Add the code related to the model implementation steps in cells inside this section

# Baseline & Model Comparisons
from sklearn.svm import LinearSVC, SVC
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, f1_score, classification_report
baseline_word = LinearSVC(C=1.0, random_state=42)
baseline_word.fit(X_train_word, y_train)
pred_word = baseline_word.predict(X_val_word)
baseline_balanced = LinearSVC(C=1.0, class_weight="balanced", random_state=42)
baseline_balanced.fit(X_train_word, y_train)
pred_balanced = baseline_balanced.predict(X_val_word)
print("Word TF-IDF Macro F1:", round(f1_score(y_val, pred_word, average="macro"), 4))
print("Balanced Macro F1:", round(f1_score(y_val, pred_balanced, average="macro"), 4))
models = {
    "Linear SVM": LinearSVC(C=0.5, random_state=42),
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Naive Bayes": MultinomialNB()
}
for name, model in models.items():
    model.fit(X_train_tfidf, y_train)
    pred = model.predict(X_val_tfidf)
    print(name, "Macro F1:", round(f1_score(y_val, pred, average="macro"), 4))
    
# Feature and Hyperparameter Adjustments
for ngram in [(2, 4), (3, 5), (3, 6), (4, 6)]:
    vec = TfidfVectorizer(analyzer="char", ngram_range=ngram, min_df=2, max_features=30000)
    x_tr = hstack([X_train_word, vec.fit_transform(X_train)])
    x_v = hstack([X_val_word, vec.transform(X_val)])
    model = LinearSVC(C=0.5, random_state=42)
    model.fit(x_tr, y_train)
    print("Character n-gram", ngram, "Macro F1:", round(f1_score(y_val, model.predict(x_v), average="macro"), 4))

for c in [0.1, 0.2, 0.3, 0.5, 0.7, 1.0, 1.5]:
    model = LinearSVC(C=c, random_state=42)
    model.fit(X_train_tfidf, y_train)
    print("C =", c, "Macro F1:", round(f1_score(y_val, model.predict(X_val_tfidf), average="macro"), 4))

for min_df in [1, 2, 3, 5]:
    vec = TfidfVectorizer(analyzer="char", ngram_range=(2, 4), min_df=min_df, max_features=30000)
    x_tr = hstack([X_train_word, vec.fit_transform(X_train)])
    x_v = hstack([X_val_word, vec.transform(X_val)])
    model = LinearSVC(C=0.5, random_state=42)
    model.fit(x_tr, y_train)
    print("min_df =", min_df, "Macro F1:", round(f1_score(y_val, model.predict(x_v), average="macro"), 4))
    
# Cross Validation and Analizer Comparison
from sklearn.model_selection import StratifiedKFold
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
texts = train_df["clean_text"].reset_index(drop=True)
labels = train_df["label"].reset_index(drop=True)
for min_df in [2, 1]:
    scores = []
    for tr_idx, val_idx in cv.split(texts, labels):
        tr_text, val_text = texts.iloc[tr_idx], texts.iloc[val_idx]
        tr_labels, val_labels = labels.iloc[tr_idx], labels.iloc[val_idx]
        w = TfidfVectorizer(ngram_range=(1, 2), min_df=2, max_features=20000)
        c = TfidfVectorizer(analyzer="char", ngram_range=(2, 4), min_df=min_df, max_features=30000)
        tr_features = hstack([w.fit_transform(tr_text), c.fit_transform(tr_text)])
        val_features = hstack([w.transform(val_text), c.transform(val_text)])
        model = LinearSVC(C=0.5, random_state=42)
        model.fit(tr_features, tr_labels)
        scores.append(f1_score(val_labels, model.predict(val_features), average="macro"))
    print("min_df =", min_df, "Fold F1:", [round(s, 4) for s in scores], "Mean:", round(np.mean(scores), 4))
for analyzer in ["char", "char_wb"]:
    vec = TfidfVectorizer(analyzer=analyzer, ngram_range=(2, 4), min_df=2, max_features=30000)
    x_tr = hstack([X_train_word, vec.fit_transform(X_train)])
    x_v = hstack([X_val_word, vec.transform(X_val)])
    model = LinearSVC(C=0.5, random_state=42)
    model.fit(x_tr, y_train)
    print(analyzer, "Macro F1:", round(f1_score(y_val, model.predict(x_v), average="macro"), 4))
    
# MiniLM Weight Tuning and V5 Model Combination Validation
mini_only = LinearSVC(C=0.5, random_state=42)
mini_only.fit(X_train_mini, y_train)
print("MiniLM only Macro F1:", round(f1_score(y_val, mini_only.predict(X_val_mini), average="macro"), 4))
for weight in [0.25, 0.5, 1.0, 1.5, 2.0, 3.0]:
    x_tr = hstack([X_train_word, X_train_char, csr_matrix(X_train_mini * weight)])
    x_v = hstack([X_val_word, X_val_char, csr_matrix(X_val_mini * weight)])
    model = LinearSVC(C=0.5, random_state=42)
    model.fit(x_tr, y_train)
    pred = model.predict(x_v)
    print("MiniLM weight =", weight, "Macro F1:", round(f1_score(y_val, pred, average="macro"), 4))
v5_model = LinearSVC(C=0.5, random_state=42)
v5_model.fit(X_train_v5, y_train)
v5_pred = v5_model.predict(X_val_v5)
print("V5 Macro F1:", f1_score(y_val, v5_pred, average="macro"))
print(classification_report(y_val, v5_pred))

# V6 RBF SVM and Decision-Score Fusion
from sklearn.preprocessing import StandardScaler
for c in [0.1, 1, 10, 100]:
    model = SVC(kernel="rbf", C=c, gamma="scale")
    model.fit(X_train_mini, y_train)
    print("RBF C =", c, "Macro F1:", round(f1_score(y_val, model.predict(X_val_mini), average="macro"), 4))
rbf_model = SVC(kernel="rbf", C=1.0, gamma="scale")
rbf_model.fit(X_train_mini, y_train)
rbf_pred = rbf_model.predict(X_val_mini)
print("Prediction agreement:", round(np.mean(v5_pred == rbf_pred), 4))
print("Both correct:", int(((v5_pred == y_val) & (rbf_pred == y_val)).sum()))
print("Only V5 correct:", int(((v5_pred == y_val) & (rbf_pred != y_val)).sum()))
print("Only RBF correct:", int(((v5_pred != y_val) & (rbf_pred == y_val)).sum()))
print("Both wrong:", int(((v5_pred != y_val) & (rbf_pred != y_val)).sum()))
assert np.array_equal(v5_model.classes_, rbf_model.classes_)
v5_scaler = StandardScaler().fit(v5_model.decision_function(X_train_v5))
rbf_scaler = StandardScaler().fit(rbf_model.decision_function(X_train_mini))
v5_scores = v5_scaler.transform(v5_model.decision_function(X_val_v5))
rbf_scores = rbf_scaler.transform(rbf_model.decision_function(X_val_mini))
for weight in [0.1, 0.2, 0.3, 0.4]:
    scores = (1 - weight) * v5_scores + weight * rbf_scores
    pred = v5_model.classes_[np.argmax(scores, axis=1)]
    print("RBF weight =", weight, "Macro F1:", round(f1_score(y_val, pred, average="macro"), 4))
word_full = TfidfVectorizer(ngram_range=(1, 2), min_df=2, max_features=20000)
char_full = TfidfVectorizer(analyzer="char", ngram_range=(2, 4), min_df=2, max_features=30000)
X_full_word = word_full.fit_transform(train_df["clean_text"])
X_test_word = word_full.transform(test_df["clean_text"])
X_full_char = char_full.fit_transform(train_df["clean_text"])
X_test_char = char_full.transform(test_df["clean_text"])
X_full_mini = normalize(minilm.encode(train_df["clean_text"].tolist(), batch_size=64, show_progress_bar=True))
X_test_mini = normalize(minilm.encode(test_df["clean_text"].tolist(), batch_size=64, show_progress_bar=True))
X_full_v5 = hstack([X_full_word, X_full_char, csr_matrix(X_full_mini * 0.5)])
X_test_v5 = hstack([X_test_word, X_test_char, csr_matrix(X_test_mini * 0.5)])
model_v5 = LinearSVC(C=0.5, random_state=42)
model_v5.fit(X_full_v5, train_df["label"])
test_pred_v5 = model_v5.predict(X_test_v5)
submission_v5 = sample[["post_id"]].merge(
    pd.DataFrame({"post_id": test_df["post_id"], "label": test_pred_v5}),
    on="post_id", how="left"
).rename(columns={"post_id": "id"})
assert submission_v5.shape == (1440, 2)
assert submission_v5["label"].isna().sum() == 0
submission_v5.to_csv("submission_v5.csv", index=False)
print("V5 submission:", submission_v5.shape)
print(submission_v5.head())

# Create Version 6 Submission and validation
rbf_v6 = SVC(kernel="rbf", C=1.0, gamma="scale")
rbf_v6.fit(X_full_mini, train_df["label"])
assert np.array_equal(model_v5.classes_, rbf_v6.classes_)
v5_scaler_v6 = StandardScaler().fit(model_v5.decision_function(X_full_v5))
rbf_scaler_v6 = StandardScaler().fit(rbf_v6.decision_function(X_full_mini))
v5_test_scores = v5_scaler_v6.transform(model_v5.decision_function(X_test_v5))
rbf_test_scores = rbf_scaler_v6.transform(rbf_v6.decision_function(X_test_mini))
final_scores = 0.8 * v5_test_scores + 0.2 * rbf_test_scores
test_pred_v6 = model_v5.classes_[np.argmax(final_scores, axis=1)]
submission_v6 = sample[["post_id"]].merge(
    pd.DataFrame({"post_id": test_df["post_id"], "label": test_pred_v6}),
    on="post_id", how="left"
).rename(columns={"post_id": "id"})
assert submission_v6.shape == (1440, 2)
assert submission_v6["label"].isna().sum() == 0
submission_v6.to_csv("submission_v6.csv", index=False)
print("V6 submission generated but not yet submitted:", submission_v6.shape)
print(submission_v6.head())
```
